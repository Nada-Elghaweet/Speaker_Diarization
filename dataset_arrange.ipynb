```python
import os

SOURCE_DIR = r"C:\Users\LaLa.STORE\OneDrive\Desktop\ANN"

print("Exists:", os.path.exists(SOURCE_DIR))
```

```python
for root, dirs, files in os.walk(SOURCE_DIR):
    print("\n Folder:", root)
    for file in files:
        print("   ", file)
```

```python
from collections import defaultdict

meetings = defaultdict(list)

for root, dirs, files in os.walk(SOURCE_DIR):
    for file in files:
        if "Headset" in file:
            meeting_id = file.split(".")[0]
            full_path = os.path.join(root, file)
            meetings[meeting_id].append(full_path)

print("Meetings found:", list(meetings.keys()))
```

```python
TARGET_DIR = r"C:\Users\LaLa.STORE\OneDrive\Desktop\ANN\AMI_cleaned"
os.makedirs(TARGET_DIR, exist_ok=True)

for meeting_id in meetings:
    os.makedirs(os.path.join(TARGET_DIR, meeting_id, "audio"), exist_ok=True)

print("Folders ready")
```

```python
import shutil

for meeting_id, file_paths in meetings.items():
    target_audio = os.path.join(TARGET_DIR, meeting_id, "audio")
    
    for src in file_paths:
        filename = os.path.basename(src)
        dst = os.path.join(target_audio, filename)
        shutil.move(src, dst)

print("Done moving")
```
